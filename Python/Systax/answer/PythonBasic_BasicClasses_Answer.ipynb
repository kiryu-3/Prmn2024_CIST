### 問題1

以下の要件を満たすクラスを作成してください

**指示**

・ moveメソッド内で「○○が歩いたよ」と出力するInsectクラスを作る  

・ moveメソッド内で「○○が飛んだよ」と出力するButterflyクラスを作る   
　 このとき、ButterflyクラスはInsectクラスを継承する  

・ moveメソッド内で「○○が跳んだよ」と出力するLocustクラスを作る   
　 このとき、LocustクラスはInsectクラスを継承する  

・ moveメソッド内で「○○が綺麗に飛んだよ」と出力するSwallowtailButterflyクラスを作る   
　 このとき、SwallowtailButterflyクラスはButterflyクラスを継承する

・ 4つのクラス全てに共通して、インスタンス化時に自分の名前を定義する。  
　このとき、コンストラクタをうまく利用する。





4つのクラスのインスタンスをリストinsect_listに格納してください。  
その後、格納した4つのインスタンスからmoveメソッドを呼び出してください。

＜実行結果例＞  
```
昆虫が歩いたよ
蝶々が飛んだよ
イナゴが跳んだよ
アゲハチョウが綺麗に飛んだよ
```

````python
# Insectクラスの定義
class Insect:
    # コンストラクタ
    def __init__(self, name):
        self.name = name  # インスタンス変数nameを定義

    # moveメソッドの定義
    def move(self):
        print(f"{self.name}が歩いたよ")

# Butterflyクラスの定義
class Butterfly:
    # コンストラクタ
    def __init__(self, name):
        self.name = name  # インスタンス変数nameを定義

    # moveメソッドの定義
    def move(self):
        print(f"{self.name}が飛んだよ")

# Locustクラスの定義
class Locust:
    # コンストラクタ
    def __init__(self, name):
        self.name = name  # インスタンス変数nameを定義

    # moveメソッドの定義
    def move(self):
        print(f"{self.name}が跳んだよ")

# SwallowtailButterflyクラスの定義
class SwallowtailButterfly:
    # コンストラクタ
    def __init__(self, name):
        self.name = name  # インスタンス変数nameを定義

    # moveメソッドの定義
    def move(self):
        print(f"{self.name}が綺麗に飛んだよ")
````

````python
# 実行するコード
# 4つのインスタンスを作成し、リストに格納
insect_list = []

insect_list.append(Insect("昆虫"))
insect_list.append(Butterfly("蝶々"))
insect_list.append(Locust("イナゴ"))
insect_list.append(SwallowtailButterfly("アゲハチョウ"))

# 各インスタンスのmoveメソッドを呼び出す
for insect in insect_list:
  insect.move()
````

### 問題2

**指示**

・ クラス名は Car、属性として変数fuelを持つ  

・ 引数なしコンストラクタでfuelを0で初期化する

・ メソッドとして戻り値、引数無しのrun() を持つ

・ run()メソッドではfuelを5消費して、走ったことが分かる文言を表示させる  
     	  　(例:  "燃料を1消費して走りました。")
				  
・ fuelが5未満の時走れなかったことが分かる文言を表示させる
        




Carに給油をするGasStationクラスを作成してください    

・ クラス名はGasStation、コンストラクタ無し

・ refuel()メソッドを持つ  

・ refuel()メソッドでは、Carクラスのインスタンスを引数として受け取る  

・ refuel()ではCarのfuelを20増やす  

・ refuel()では給油したことが分かる文言を表示する

CarとGasStationをインスタンス化し、  
GasStationで給油をする前と後でCarのrun()を呼び出してください

＜実行結果例＞

```
燃料が足りないため走れませんでした。
給油したことによりfuelが20増えました。
燃料を1消費して走りました。
```

````python
class Car:
    def __init__(self):
        self.fuel = 0  # 燃料の初期値を設定

    def run(self):
        # 燃料が5未満の場合は走れないことを表示
        if self.fuel < 5:
            print("燃料が足りないため走れませんでした。")
        else:
            # 燃料を5消費して走ったことを表示
            print("燃料を5消費して走りました。")
            self.fuel -= 5  # 燃料を5消費する
````

````python
class GasStation:
    def refuel(self, car):
        car.fuel += 20  # 車の燃料を20増やす
        # 給油したことにより燃料が20増えたことを表示
        print(f"給油したことにより、fuelが20増えました。")
````

````python
# CarとGasStationのインスタンスを作成して、動作確認を行う
car = Car()
gasstation = GasStation()

car.run()  # 車を走行させる
gasstation.refuel(car)  # ガソリンスタンドで給油する
car.run()  # 再び車を走行させる
````

### 問題2.5（問題2＋アルファ）

問題2を拡張してください。



・ Carクラスでは、コンストラクタで車の最大燃料容量を設定する。

・ Carクラスでは、車の現在の燃料残量を表示するdisplay_fuel()メソッドを追加する。

・ GasStationクラスでは、給油時に最大燃料容量を超えないように制限する。  
　  最大燃料容量までは燃料を給油できるようにする。

＜実行結果例＞

```
燃料が足りないため走れませんでした。
給油したことにより、fuelが20増えました。
燃料を5消費して走りました。
現在の燃料残量: 15/50
給油したことにより、fuelが20増えました。
給油したことにより、fuelが15増えました。
燃料タンクが満タンです。
現在の燃料残量: 50/50
```

````python
class Car:
    def __init__(self, max_fuel):
        self.fuel = 0  # 燃料の初期値を設定
        self.max_fuel = max_fuel  # 最大燃料容量を設定

    def run(self):
        # 燃料が5未満の場合は走れないことを表示
        if self.fuel < 5:
            print("燃料が足りないため走れませんでした。")
        else:
            # 燃料を5消費して走ったことを表示
            print("燃料を5消費して走りました。")
            self.fuel -= 5  # 燃料を5消費する

    def display_fuel(self):
        # 現在の燃料残量と最大燃料容量を表示
        print(f"現在の燃料残量: {self.fuel}/{self.max_fuel}")
````

````python
class GasStation:
    def refuel(self, car):
        if car.fuel + 20 <= car.max_fuel:  # 燃料を給油しても最大容量を超えない場合
            car.fuel += 20
            print(f"給油したことにより、fuelが20増えました。")
        else:  # 燃料を給油しても最大容量を超える場合
            print(f"給油したことにより、fuelが{car.max_fuel - car.fuel}増えました。")
            car.fuel += (car.max_fuel - car.fuel)  # 燃料を最大容量まで給油
            print(f"燃料タンクが満タンです。")
````

````python
# CarとGasStationのインスタンスを作成して、動作確認を行う
car = Car(50)
gasstation = GasStation()

car.run()  # 車を走行させる
gasstation.refuel(car)  # ガソリンスタンドで給油する
car.run()  # 再び車を走行させる
car.display_fuel()  # 現在の燃料残量を表示
gasstation.refuel(car)  # ガソリンスタンドで再度給油する
gasstation.refuel(car)  # ガソリンスタンドで給油する
car.display_fuel()  # 現在の燃料残量を表示
````

**指示**

口座情報を管理するAccountクラスを作成してください

銀行でのお金のやり取りを行うATMクラスを作成してください  

・ ATMのコンストラクタでAccountのインスタンスを管理するaccountsディクショナリを初期化する  
　・ キーは口座番号、値をインスタンスにする

・ register_account()メソッドではアカウントを登録する

 　・ このとき、すでに使われているnumberで登録しようとしていたら、

・ exists_account()メソッドは引数のnumberから   accountsに存在するかどうかをbooleanで返す

 　 残高が足りないときには行えないようにする
        


それぞれ何か行うたびに、何をしたか分かるように標準出力で表示するようにしてください

＜実行結果例＞

```

佐藤太郎 さんのアカウントを口座番号:12345 で登録しました。
```

````python
class Account:
    def __init__(self, name, number):
        self.name = name  # 口座名義人の名前を設定
        self.number = number  # 口座番号を設定
        self.balance = 0  # 口座残高を初期化
````

````python
class ATM:
    def __init__(self):
        self.accounts = dict()  # 口座情報を格納する辞書を初期化

    # 口座登録メソッド
    def register_account(self, name, number):
        # 口座が1つ以上存在する場合
        if len(list(self.accounts.keys())) != 0:
            # 登録しようとしている口座番号が既に存在する場合
            if number in list(self.accounts.keys()):
                print(f"口座番号:{number} は 名前:{name} さんの口座です。")
            else:
                # 口座を新規登録
                self.accounts[f"{number}"] = Account(name, number)
                print(f"{name} さんのアカウントを口座番号:{number} で登録しました。")
        else:
            # 口座を新規登録
            self.accounts[f"{number}"] = Account(name, number)
            print(f"{name} さんのアカウントを口座番号:{number} で登録しました。")

    # 口座存在確認メソッド
    def exists_account(self, name, number):
        # 口座番号が存在する場合
        if number in list(self.accounts.keys()):
            print(f"口座番号:{number} は 名前:{name} さんの口座です。")
            return True
        else:
            # 口座番号が存在しない場合
            print(f"口座番号:{number} は存在しません。")
            return False

    # 入金メソッド
    def deposit(self, number, money):
        # 口座番号が存在する場合
        if number in list(self.accounts.keys()):
            # 入金処理
            print(f"口座番号:{number} に{money} 円入金しました。")
            self.accounts[f'{number}'].balance += money
        else:
            # 口座番号が存在しない場合
            print(f"口座番号:{number}は存在しませんでした。")

    # 引き出しメソッド
    def withdraw(self, number, money):
        # 口座番号が存在する場合
        if number in list(self.accounts.keys()):
            # 引き出し可能かチェック
            if self.accounts[f'{number}'].balance - money < 0:
                print(f"口座番号:{number} から{money} 円引き出せませんでした。", end="")
                print(f"残高:{self.accounts[f'{number}'].balance} 円です。")
            else:
                # 引き出し処理
                print(f"口座番号:{number} から{money} 円引き出しました。", end="")
                print(f"残高:{self.accounts[f'{number}'].balance - money} 円です。")
                self.accounts[f'{number}'].balance -= money
        else:
            # 口座番号が存在しない場合
            print(f"口座番号:{number} は存在しませんでした。")
````

````python
# ATMとアカウントの操作
atm = ATM()

# 佐藤太郎のアカウントが存在しない場合、登録
atm.exists_account("佐藤太郎", "12345")
atm.register_account("佐藤太郎", "12345")

# 佐藤次郎のアカウントが存在しない場合、登録
atm.register_account("佐藤次郎", "12346")

# 佐藤太郎のアカウントが存在する場合、入金と引き出しを実施
if atm.exists_account("佐藤太郎", "12345"):
    atm.deposit("12345", 1000)
    atm.withdraw("12345", 2000)
    atm.withdraw("12345", 500)

# 佐藤次郎のアカウントが存在する場合、入金と引き出しを実施
if atm.exists_account("佐藤次郎", "12346"):
    atm.deposit("12346", 2000)
    atm.withdraw("12346", 1000)
    atm.withdraw("12346", 1500)
````

### 問題3.5（問題3＋アルファ）

問題3を拡張してください。



・ Accountクラスにtransactionsというリスト属性を追加し、取引の履歴をリストで保持する。  

・ 取引を行うたびに、取引の詳細（日付、金額、取引の種類）をtransactionsリストに追加する。  
　日付はコードが実行されたときの日時を自動で取得する

・ show_transaction_history()メソッドを作成し、これまでの取引履歴（日付、金額、取引の種類）を表示する。  
　表示する履歴は最近のもの10件とする。

＜実行結果例＞

```
口座番号:12345 は存在しません.
佐藤太郎 さんのアカウントを口座番号:12345 で登録しました.
口座番号:12345 は 名前:佐藤太郎 さんの口座です.
口座番号:12345 に1000 円入金しました.
口座番号:12345 から2000 円引き出せませんでした. 残高:1000 円です.
口座番号:12345 から500 円引き出しました. 残高:500 円です.
取引履歴:
2024-02-18 18:50:35: 入金 1000 円
2024-02-18 18:50:35: 引出し 500 円

```

````python
class Account:
    def __init__(self, name, number):
        self.name = name
        self.number = number
        self.balance = 0
        self.transactions = []
````

````python
from datetime import datetime, timezone, timedelta

class ATM:
    def __init__(self):
        self.accounts = dict()

    # アカウントの登録
    def register_account(self, name, number):
        # すでに登録されているアカウントの場合
        if number in self.accounts:
            print(f"口座番号:{number} は 名前:{name} さんの口座です。")
        else:
            self.accounts[number] = Account(name, number)
            print(f"{name} さんのアカウントを口座番号:{number} で登録しました.")

    # アカウントの存在確認
    def exists_account(self, name, number):
        if number in self.accounts:
            print(f"口座番号:{number} は 名前:{name} さんの口座です.")
            return True
        else:
            print(f"口座番号:{number} は存在しません.")
            return False

    # 入金
    def deposit(self, number, money):
        if number in self.accounts:
            self.accounts[number].balance += money
            print(f"口座番号:{number} に{money} 円入金しました.")
            self.accounts[number].transactions.append(self.add_transaction("入金", money))
        else:
            print(f"口座番号:{number} は存在しませんでした.")

    # 引き出し
    def withdraw(self, number, money):
        if number in self.accounts:
            # 引き出し可能な場合
            if self.accounts[number].balance - money >= 0:
                self.accounts[number].balance -= money
                print(f"口座番号:{number} から{money} 円引き出しました. 残高:{self.accounts[number].balance} 円です.")
                self.accounts[number].transactions.append(self.add_transaction("引出し", money))
            else:
                print(f"口座番号:{number} から{money} 円引き出せませんでした. 残高:{self.accounts[number].balance} 円です.")
        else:
            print(f"口座番号:{number} は存在しません.")

    # 取引履歴を追加
    def add_transaction(self, transaction_type, amount):
        jst = timezone(timedelta(hours=+9), 'JST')
        current_time = datetime.now(jst)
        formatted_time = current_time.strftime("%Y-%m-%d %H:%M:%S")  # 日本の現在時刻をフォーマット
        transaction_info = f"{formatted_time}: {transaction_type} {amount} 円"
        return transaction_info

    # 取引履歴を表示
    def show_transaction_history(self, number):
        if number in self.accounts:
            print("取引履歴:")
            if len(self.accounts[number].transactions) > 10:
                recent_transactions = self.accounts[number].transactions[-10:]
                for transaction in recent_transactions:
                    print(transaction)
            else:
                for transaction in self.accounts[number].transactions:
                    print(transaction)
        else:
            print(f"口座番号:{number} は存在しませんでした.")
````

````python
# ATMとアカウントの操作
atm = ATM()

# 佐藤太郎のアカウントが存在しない場合、登録
atm.exists_account("佐藤太郎", "12345")
atm.register_account("佐藤太郎", "12345")

# 佐藤太郎のアカウントが存在する場合、入金と引き出しを実施
if atm.exists_account("佐藤太郎", "12345"):
    atm.deposit("12345", 1000)
    atm.withdraw("12345", 2000)
    atm.withdraw("12345", 500)
    atm.show_transaction_history("12345")
````
