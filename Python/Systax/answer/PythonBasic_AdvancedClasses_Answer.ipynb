### 問題1

以下の要件を満たすクラスを作成してください

**指示**

・ moveメソッド内で「○○が歩いたよ」と出力するInsectクラスを作る  

・ moveメソッド内で「○○が飛んだよ」と出力するButterflyクラスを作る   
　 このとき、ButterflyクラスはInsectクラスを継承する  

・ moveメソッド内で「○○が跳んだよ」と出力するLocustクラスを作る   
　 このとき、LocustクラスはInsectクラスを継承する  

・ moveメソッド内で「○○が綺麗に飛んだよ」と出力するSwallowtailButterflyクラスを作る   
　 このとき、SwallowtailButterflyクラスはButterflyクラスを継承する

・ 4つのクラス全てに共通して、インスタンス化時に自分の名前を定義する。  
　このとき、コンストラクタをうまく利用する。





4つのクラスのインスタンスをリストinsect_listに格納してください。  
その後、格納した4つのインスタンスからmoveメソッドを呼び出してください。

````python
class Insect:
  def __init__(self, name):
    self.name = name

  def move(self):
    print(f"{self.name}が歩いたよ")

class Butterfly(Insect):
  # サブクラスのメソッドが優先される
  def move(self):
    print(f"{self.name}が飛んだよ")

class Locust(Insect):
  # サブクラスのメソッドが優先される
  def move(self):
    print(f"{self.name}が跳んだよ")

class SwallowtailButterfly(Butterfly):
  # サブクラスのメソッドが優先される
  def move(self):
    print(f"{self.name}が綺麗に飛んだよ")
````

````python
# 実行するコード
# 4つのインスタンスを作成し、リストに格納
insect_list = []

insect_list.append(Insect("昆虫"))
insect_list.append(Butterfly("蝶々"))
insect_list.append(Locust("イナゴ"))
insect_list.append(SwallowtailButterfly("アゲハチョウ"))

# 各インスタンスのmoveメソッドを呼び出す
for insect in insect_list:
  insect.move()
````

### 問題2

以下の要件を満たすクラスを作成してください

**指示**

・ クラス名は Car、属性として変数fuelを持つ  

・ 引数なしコンストラクタでfuelを0で初期化する

・ メソッドとして戻り値、引数無しのrun() を持つ

・ run()メソッドではfuelを5消費して、走ったことが分かる文言を表示させる  
     	  　(例:  "燃料を1消費して走りました。")
				  
・ fuelが5未満の時走れなかったことが分かる文言を表示させる
        




Carに給油をするGasStationクラスを作成してください    

・ クラス名はGasStation、コンストラクタ無し

・ refuel()メソッドを持つ  

・ refuel()メソッドでは、Carクラスのインスタンスを引数として受け取る  

・ refuel()ではCarのfuelを20増やす  

・ refuel()では給油したことが分かる文言を表示する

今回はCarクラスを継承したTruckクラスを作っていきましょう。

**指示**

・ クラス名は Truck、Carクラスを継承する

・ Carクラスの属性に加えて、定格積載量を表す属性cargo_capacity（コンストラクタで定義）と、  
 　 現在の乗員数を表す属性cargo（最初は0kg）を定義する

・ load_cargo()メソッドで、指定した量の荷物をトラックに積んだり降ろしたりできるようにする    
  （積載量がマイナスにならないように注意する）

・ 正の値を指定した場合はトラックに乗せ、負の値を指定した場合はトラックから降ろす  

・ 定格積載量をオーバーしたらその旨を出力する。だが、荷物をさらに積むことは可能にする

・ show_cargo()メソッドで、現在の荷物の重量を確認できるようにする
        

＜実行結果例＞  
```
燃料が足りないため走れませんでした。
給油したことにより、fuelが20増えました。
燃料を5消費して走りました。
20 kgの荷物を積みました。
60 kgの荷物を積みました。定格積載量を超えています。
燃料を5消費して走りました。
40 kgの荷物を降ろしました。
燃料を5消費して走りました。
現在の荷物の重量は40 kgです。
```

````python
class Car:
    def __init__(self):
        self.fuel = 0

    def run(self):
        # 燃料が5未満の場合は走行できない
        if self.fuel < 5:
            print("燃料が足りないため走れませんでした。")
        else:
            print("燃料を5消費して走りました。")
            self.fuel -= 5
````

````python
class GasStation:
    def refuel(self, car):
        # ガソリンスタンドからの給油処理
        self.fuel = 20
        car.fuel += self.fuel
        print(f"給油したことにより、fuelが{self.fuel}増えました。")
````

````python
class Truck(Car):
    def __init__(self, cargo_capacity):
        super().__init__()  # スーパークラスのコンストラクタを呼び出す
        self.cargo_capacity = cargo_capacity
        self.cargo = 0

    def load_cargo(self, amount):
        # 荷物の積み込みおよび積み下ろし処理
        if amount > 0:
            if self.cargo + amount <= self.cargo_capacity:
                self.cargo += amount
                print(f"{amount} kgの荷物を積みました。")
            else:
                self.cargo += amount
                print(f"{amount} kgの荷物を積みました。定格積載量を超えています。")
        else:
            if self.cargo >= abs(amount):
                self.cargo += amount
                print(f"{abs(amount)} kgの荷物を降ろしました。")
            else:
                print("荷物を降ろすことができません。荷物が不足しています。")

    def show_cargo(self):
        # 現在の荷物の重量を表示
        print(f"現在の荷物の重量は{self.cargo} kgです。")
````

````python
# トラックのインスタンス化と動作確認
truck = Truck(50)
gasstation = GasStation()

truck.run()  # 燃料が不足しているため走行できない
gasstation.refuel(truck)  # ガソリンスタンドから給油
truck.run()  # 燃料を消費して走行

truck.load_cargo(20)  # 荷物を積む
truck.load_cargo(60)  # 定格積載量を超える量の荷物を積む
truck.run()  # 燃料を消費して走行
truck.load_cargo(-40)  # 荷物を降ろす
truck.run()  # 燃料を消費して走行
truck.show_cargo()  # 現在の荷物の重量を表示
````

### 問題3

以下の要件を満たすクラスを作成してください

**指示**

 Car クラスのrun()メソッドをTruckクラス内でオーバーライドしてください  

・ 積載量が定格積載量を超えていたらバスが動けない旨を出力する  

・ 何kg降ろせば動けるようになるか（定格積載量まで積載量を減らせるか）も同時に出力する


＜実行結果例＞

```
燃料が足りないため走れませんでした。
給油したことにより、fuelが20増えました。
燃料を5消費して走りました。
20 kgの荷物を積みました。
60 kgの荷物を積みました。定格積載量を超えています。
定格積載量なので走れません。最低でも30kg降ろす必要があります。
40 kgの荷物を降ろしました。
燃料を5消費して走りました。
現在の荷物の重量は40 kgです。
```

````python
class Truck(Car):
    def __init__(self, cargo_capacity):
        super().__init__()  # スーパークラスのコンストラクタを呼び出す
        self.cargo_capacity = cargo_capacity
        self.cargo = 0

    # Carクラスのrun()メソッドをオーバーライド
    def run(self):
        # 積載量が定格積載量を超える場合は走行できない
        if self.cargo > self.cargo_capacity:
            print(f"定格積載量なので走れません。最低でも{self.cargo - self.cargo_capacity}kg降ろす必要があります。")
        else:
            super().run()

    def load_cargo(self, amount):
        # 荷物の積み込みおよび積み下ろし処理
        if amount > 0:
            if self.cargo + amount <= self.cargo_capacity:
                self.cargo += amount
                print(f"{amount} kgの荷物を積みました。")
            else:
                self.cargo += amount
                print(f"{amount} kgの荷物を積みました。定格積載量を超えています。")
        else:
            if self.cargo >= abs(amount):
                self.cargo += amount
                print(f"{abs(amount)} kgの荷物を降ろしました。")
            else:
                print("荷物を降ろすことができません。荷物が不足しています。")

    def show_cargo(self):
        # 現在の荷物の重量を表示
        print(f"現在の荷物の重量は{self.cargo} kgです。")
````

````python
# トラックのインスタンス化と動作確認
truck = Truck(50)
gasstation = GasStation()

truck.run()  # 走行できない旨のメッセージが表示される
gasstation.refuel(truck)  # ガソリンスタンドから給油
truck.run()  # 燃料を消費して走行

truck.load_cargo(20)  # 荷物を積む
truck.load_cargo(60)  # 定格積載量を超える量の荷物を積む
truck.run()  # 燃料を消費して走行
truck.load_cargo(-40)  # 荷物を降ろす
truck.run()  # 燃料を消費して走行
truck.show_cargo()  # 現在の荷物の重量を表示
````

### 問題4（問題2・3＋アルファ）

今回はTruckクラスを継承したPickupTruckクラスを作っていきましょう。

※オフロードモード：オフロード（滑りやすい路面）走行時の駆動輪の空転を防ぎ､走破性を高めるモード。

**指示**

・ クラス名は PickupTruck、Truckクラスを継承する

・ Truckクラスの属性に加えて、オフロードモードかどうか表す属性offroad_mode（コンストラクタで定義）を定義する

・ load_cargo()メソッドで、オフロードモードではなくまた荷物を積もうとしているときに限り、  
　オフロードモードではないため荷物を積めない旨を出力する。  
　（Truckクラスのload_cargo()メソッドをオーバーライド）

・ show_offroad_mode()メソッドで、現在のオフロードモードの状態を確認できるようにする。

・ change_offroad_mode()メソッドで、オフロードモードの状態を切り替える。

        

＜実行結果例＞

```
燃料が足りないため走れませんでした。
給油したことにより、fuelが20増えました。
燃料を5消費して走りました。
オフロードモードではないため、荷物を積みません。
オフロードモードは無効です。
オフロードモードが有効です。
20 kgの荷物を積みました。
60 kgの荷物を積みました。定格積載量を超えています。
定格積載量なので走れません。最低でも30kg降ろす必要があります。
40 kgの荷物を降ろしました。
燃料を5消費して走りました。
現在の荷物の重量は40 kgです。
```

````python
class PickupTruck(Truck):
    def __init__(self, cargo_capacity):
        super().__init__(cargo_capacity)
        self.offroad_mode = False

    def load_cargo(self, amount):
        # オフロードモードでない場合は荷物を積まない
        if not self.offroad_mode and amount >= 0:
            print("オフロードモードではないため、荷物を積みません。")
        else:
            super().load_cargo(amount)

    def show_offroad_mode(self):
        # オフロードモードの状態を表示
        if self.offroad_mode:
            print("オフロードモードが有効です。")
        else:
            print("オフロードモードは無効です。")

    def change_offroad_mode(self):
        # オフロードモードの切り替え
        self.offroad_mode = not self.offroad_mode
````

````python
# PickupTruckのインスタンス化と動作確認
pickuptruck = PickupTruck(50)
gasstation = GasStation()

pickuptruck.run()  # 走行
gasstation.refuel(pickuptruck)  # 給油
pickuptruck.run()  # 走行

pickuptruck.load_cargo(20)  # 荷物を積む
pickuptruck.show_offroad_mode()  # オフロードモードの状態を表示
pickuptruck.change_offroad_mode()  # オフロードモードの切り替え
pickuptruck.show_offroad_mode()  # オフロードモードの状態を表示
pickuptruck.load_cargo(20)  # 荷物を積む（オフロードモードなので積む）
pickuptruck.load_cargo(60)  # 荷物を積む（定格積載量を超えるため積まない）
pickuptruck.run()  # 走行

pickuptruck.load_cargo(-40)  # 荷物を降ろす
pickuptruck.run()  # 走行
pickuptruck.show_cargo()  # 現在の荷物の重量を表示
````
