````python
# !!WRITE ME!!に処理を記入する
def homework(file_path):
    # テキストファイルからデータを読み込む
    with open(file_path, mode="r") as f:
        lines = f.readlines()

    # ヘッダーを取得し、カンマで分割してキーとして使用する
    header = lines[0].strip().split(",")

    # データを指定された形式に変換する
    formatted_data = []
    for line in lines[1:]:
        values = line.strip().split(",")
        data = {header[i]: values[i] for i in range(len(header))}
        formatted_data.append(data)

    # フォーマットされたデータを出力する
    for data in formatted_data:
        print(data)

homework("household_expenses.txt")
````

GPAの計算式は以下の通りです。  
$$

````python
def homework(file_path):
    # 成績を GPA に変換する関数
    def convert_to_GP(grade):
        if grade == "秀":
            return 4
        elif grade == "優":
            return 3
        elif grade == "良":
            return 2
        elif grade == "可":
            return 1
        else:
            return 0

    # GPA を計算する関数
    def calculate_GPA(subjects, grades, one_unit_class):
        total_gp_units = 0  # 総合 GPA 単位
        total_units = 0  # 総合取得単位数

        # 各科目について GPA を計算
        for subject, grade in zip(subjects, grades):
            # 一単位科目かどうかで GPA 単位を決定
            if subject in one_unit_class:
                gp_units = convert_to_GP(grade) * 1
            else:
                gp_units = convert_to_GP(grade) * 2
            total_gp_units += gp_units
            total_units += 1 if subject in one_unit_class else 2

        # GPA を計算
        gpa = total_gp_units / total_units
        return gpa

    # テキストファイルからデータを読み取る関数
    def text_read(file_path):
        with open(file_path, mode="r+") as f:
            lines = f.read().splitlines()

        seiseki_data = {"授業": [], "成績": []}  # 科目と成績の辞書
        one_unit_class = []  # 一単位科目のリスト
        for i in range(1, len(lines)):
            kamoku = lines[i].split("\t")[3]  # 科目名
            score = lines[i].split("\t")[4]  # 成績

            # 科目と成績を辞書に追加
            seiseki_data["授業"].append(kamoku)
            seiseki_data["成績"].append(score)

            # 一単位科目のリストに追加
            if lines[i].split("\t")[5] == "1":
                one_unit_class.append(kamoku)

        return seiseki_data, one_unit_class

    # テキストファイルからデータを読み取る
    seiseki_data, one_unit_class = text_read(file_path)
    subjects = seiseki_data["授業"]  # 科目リスト
    grades = seiseki_data["成績"]  # 成績リスト
    gpa = calculate_GPA(subjects, grades, one_unit_class)  # GPA を計算

    # GPA を出力
    print(f"GPAは{gpa:.2f}です")


homework("seiseki.txt")
````

データの取得・テキストファイルの書き込みは以下のコードを実行してください。

````python
import os
import glob
import re

# gitリポジトリからKyotoCorpusをクローンする
!git clone https://github.com/ku-nlp/KyotoCorpus

# KyotoCorpusからnumファイルを取得する
files = glob.glob("KyotoCorpus/dat/num/*")

# ひらがなを抽出する正規表現パターン
noun_regrex = re.compile(r'''(
    [\u3040-\u309F]+  # ひらがなのUnicode範囲
)''', re.VERBOSE)

# 抽出されたひらがなのリスト
rep_list = []

# 各ファイルを読み込んで名詞を抽出する
for filename in files:
    with open(filename, "r") as f:
        for line in f:
            # 名詞と地名、組織名、固有名詞、普通名詞を含む行を抽出
            if "名詞" in line and ("地名" in line or "組織名" in line or "固有名詞" in line or "普通名詞" in line):
                # ひらがなを抽出
                rep = noun_regrex.findall(line)
                # 抽出結果がある場合はリストに追加
                if rep:
                    word = rep[0]
                    # 'っ'、'ゔ'、'ぢ'から始まる単語でない場合のみ追加
                    if not word.startswith('っ') and not word.startswith('ゔ') and not word.startswith('ぢ'):
                        rep_list.append(word)

# 重複を取り除く
rep_list = list(set(rep_list))
# ひらがなの順にソート
rep_list.sort()

# output.txtに書き込む
with open("output.txt", "w", encoding='UTF-8') as output:
    for rep in rep_list:
        output.write(rep + "\n")
````
