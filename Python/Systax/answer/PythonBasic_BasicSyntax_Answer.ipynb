### 問題1

「*」を使って、指定された段数のひし形を作成する関数`homework`を作成してください。  
関数`homework`は引数`length`を受け取り、`length`がひし形の高さ（中央の列の高さ）を表します。

以下は`homework`関数の利用例です。

``` python
homework(4)
```

出力

``` markdown
    *
   ***
  *****
 *******
  *****
   ***
    *
```


````python
# 解答例のコード
def homework(length):
    for i in range(1, 2 * length + 1):
        N1 = abs(length - i)      # 空白の数（片方）
        N2 = (2 * length - 1) - 2 * N1  # "*"の数

        for j in range(N1):
            print(" ", end="")

        for j in range(N2):
            print("*", end="")

        print()  # 改行
````

### 問題2

「*」を使って、指定された段数のひし形を2個横に並べて作成する関数`homework`を作成してください。  
関数`homework`は引数`length`を受け取り、`length`がひし形の高さ（中央の列の高さ）を表します。

以下は`homework`関数の利用例です。

``` python
homework(4)
```

出力

``` markdown
   *     *
  ***  ***
 ***** *****
*************

```


````python
# 解答例のコード
def homework(length):
    for i in range(1, 2 * length):
        N1 = abs(length - i)  # 空白の数（左方）
        N2 = (2 * length - 1) - 2 * N1  # "*"の数（左方）

        if i == length:
            N3 = 2 * N1  # 空白の数（真ん中）
            N4 = N2 - 1  # "*"の数（右方）
        else:
            N3 = 2 * N1 - 1  # 空白の数（真ん中）
            N4 = N2  # "*"の数（右方）

        for j in range(N1):
            print(" ", end="")

        for j in range(N2):
            print("*", end="")

        for j in range(N3):
            print(" ", end="")

        for j in range(N4):
            print("*", end="")

        print()  # 改行
````

````python
# 解答例のコード
def homework(length):
    max_length = (2 * length - 1) * 2 - 1  # 真ん中の行の長さ

    for i in range(1, 2 * length):
        N1 = abs(length - i)  # 空白の数（左方）
        N2 = (2 * length - 1) - 2 * N1  # 真ん中の行の長さ - 空白の数（両方）
        N3 = max_length - (N1 * 2 + N2 * 2) + int(i == length)  # 空白の数（真ん中）
        N4 = max_length - (N1 * 2 + N2 + N3)  # "*"の数（右方）

        print(" " * N1 + "*" * N2 + " " * N3 + "*" * N4)
````

### 問題4

以下のデータのGPAを算出する関数`homework`を作成してください。  
関数`homework`は引数`seiseki_data`と`one_unit_class`を受け取ります。

- `seiseki_data`：成績データ
- `one_unit_class`：一単位の科目。そのほかの単位は二単位。


````python
seiseki_data = {
    "授業": ["物理学入門", "化学入門", "電子工学入門","情報演習", "キャリア入門",
             "地域学入門","言語リテラシー入門", "微分積分学入門", "工学実験入門","理系英語入門",
             "日本国憲法入門", "スポーツ科学入門", "健康学入門"],
    "成績": ["良", "秀", "秀", "秀", "秀", "秀", "秀", "優", "優", "秀", "秀", "優", "優"]
}

one_unit_class = ("キャリア入門", "地域学入門", "言語リテラシー入門",
                  "工学実験入門", "スポーツ科学入門", "健康学入門")
````

成績データは、科目名とその科目の成績の組み合わせで構成されます。  
成績は、次の表に示すように、5段階評価（秀、優、良、可、不可）に対応するGP（Grade Point）で表されます。

- 秀: 4
- 優: 3
- 良: 2
- 可: 1
- 不可: 0

GPAの計算式は以下の通りです。  
$$
GPA = \frac{(GP×そのGPの単位数) の合計}{履修単位数}
$$

````python
def homework(seiseki_data, one_unit_class):
    # 成績を GPA に変換する関数
    def convert_to_GP(grade):
        if grade == "秀":
            return 4
        elif grade == "優":
            return 3
        elif grade == "良":
            return 2
        elif grade == "可":
            return 1
        else:
            return 0

    # GPA を計算する関数
    def calculate_GPA(subjects, grades, one_unit_class):
        total_gp_units = 0  # 総合 GPA 単位
        total_units = 0  # 総合取得単位数

        # 各科目について GPA を計算
        for subject, grade in zip(subjects, grades):
            # 一単位科目かどうかで GPA 単位を決定
            if subject in one_unit_class:
                gp_units = convert_to_GP(grade) * 1
            else:
                gp_units = convert_to_GP(grade) * 2
            total_gp_units += gp_units
            total_units += 1 if subject in one_unit_class else 2

        # GPA を計算
        gpa = total_gp_units / total_units
        return gpa

    # 科目と成績データを取得
    subjects = seiseki_data["授業"]
    grades = seiseki_data["成績"]

    # GPA を計算
    gpa = calculate_GPA(subjects, grades, one_unit_class)

    # GPA を出力
    print(f"GPAは{gpa:.2f}です")
````

````python
homework(seiseki_data, one_unit_class)
````

### 問題5

**FizzBuzz（フィズ・バズ）問題**という問題について解いていきます。

もともとは、英語圏で長距離ドライブ中や飲み会の時に行われる言葉遊びです。

1～Nまでの整数を順に出力しなさい。ただし、例外として以下のルールを守ること。

* 3で割り切れる数字の場合はFizzを出力
* 5で割り切れる数字の場合はBuzzを出力
* 7で割り切れる数字の場合はJazzを出力
* 複数条件が当てはまる場合は組み合わせて出力（ "FizzBuzz", "FizzJazz", "BuzzJazz" など）

"Fizz" か "Buzz" か "Jazz" か "組み合わさったもの" か "others"、  
それぞれの要素の数をカウントし、ディクショナリの形で出力する関数`homework`を作成してください。  
関数`homework`は引数`n`を受け取り、`n`が整数範囲の最大値を表します。

＜実行結果例＞  
``` py
{'Fizz': 23, 'Buzz': 12, 'Jazz': 8, 'FizzBuzz': 6, 'FizzJazz': 4, 'BuzzJazz': 2, 'others': 45}
```

````python
def homework(n):
    # 結果を格納するリスト
    result = []

    # FizzBuzzルールに基づいて結果を生成
    for i in range(1, n+1):
        output = ""
        if i % 3 == 0:
            output = "Fizz"
        if i % 5 == 0:
            output += "Buzz"
        if i % 7 == 0:
            output += "Jazz"

        # 結果が空の場合は"others"を追加
        if not output:
            result.append("others")
        else:
            result.append(output)

    # 結果の出現回数を数える
    FizzBuzz_dic = {}
    for key in set(result):
        FizzBuzz_dic[key] = result.count(key)

    # ディクショナリの順番を修正する
    keys_order = ['Fizz', 'Buzz', 'Jazz', 'FizzBuzz', 'FizzJazz', 'BuzzJazz', 'others']
    ordered_result = {key: FizzBuzz_dic[key] for key in keys_order}

    # ソートされた結果を出力
    print(ordered_result)
````

````python
homework(100)
````
